In this notebook, we use the Vinted API and compare its Synchrone and Asynchrone usage. The synchrone setup performs task one by one, waiting for the first task to be finished to start another one. The synchrone setup does not consume a lot of CPU but is slower than the asynchrone setup. The asynchrone setup allows multiple research tasks at once and consumes a bit more CPU. Requests are considered in parallel, producing lower time lapse. For a large number of requests, the asynchrone setup is best suited. Fot that start, no proxie has been implemented but for further improvement, proxie should be considered to allow multiple parallel requests in parallel to the Vinted API. 

## Synchrone setup

```python
from vinted_scraper import VintedScraper
# Initialize scraper with base URL
scraper = VintedScraper("https://www.vinted.com") #(domain="fr") for french market
# Define search parameters
params = {"search_text": "board games"}
# price_from, price_to 

# Perform search - returns List[VintedItem]
items = scraper.search(params)
```

## Asynchrone setup

```python
from vinted_scraper import AsyncVintedScraper

# Initialize async scraper using factory method
scraper = await AsyncVintedScraper.create("https://www.vinted.com")

# Define search parameters
params = {
    "search_text": "board games",
    "price_to": 50,
    "order": "newest_first" # price_low_to_high, relevance, price_high_to_low
        }

# Perform async search - returns List[VintedItem]
items = await scraper.search(params)

# Enrich first item with description from the item page
if items:
    first = items[0]
    await scraper.enrich(first)
    print(f"Title: {first.title}")
    print(f"Description: {first.description}")
```
